```sql
create catalog if not exists IDENTIFIER(:catalog);
create schema if not exists IDENTIFIER(:catalog || '.raw');
create schema if not exists identifier(:catalog || '.bronze');
create schema if not exists identifier(:catalog || '.silver');
create schema if not exists identifier(:catalog || '.gold');
create volume if not exists identifier(:catalog || '.raw.landing');
```

```sql
create table if not exists identifier(:catalog || '.raw.plans') as
select *
from read_files('/Volumes/claims/raw/landing/raw/plans.csv')
```

```sql
create table if not exists identifier(:catalog || '.raw.hospitals') as
select *
from read_files('/Volumes/claims/raw/landing/raw/hospitals.csv')
```
